```python
import pandas as pd

df = pd.read_csv('ohlc_clean.csv')
df['date'] = pd.to_datetime(df['date'])

# Keep only close, pivot to wide format (rows=date, columns=ticker)
wide = df.pivot(index='date', columns='ticker', values='close')

# Daily return % — (today - yesterday) / yesterday * 100
returns = wide.pct_change() * 100
returns = returns.round(4)

# Drop first row (all NaN — no previous day)
returns = returns.dropna(how='all')

returns.index.name = 'date'
returns.to_csv('ohlc_processed.csv')

print(f"Saved: ohlc_processed.csv — {returns.shape[0]} dates × {returns.shape[1]} tickers")
print(f"Date range: {returns.index[0].date()} → {returns.index[-1].date()}")
print(f"Sample:\n{returns.iloc[:3, :5]}")
```
