```python
import yfinance as yf
import pandas as pd

tickers = {

    # Oil & Gas — OMCs
    "HPCL":              "HINDPETRO.NS",
    "BPCL":              "BPCL.NS",
    "Indian Oil":        "IOC.NS",
    "Indraprastha Gas":  "IGL.NS",
    "Mahanagar Gas":     "MGL.NS",
    "Gujarat Gas":       "GUJARATGAS.NS",

    # Oil & Gas — Upstream
    "ONGC":              "ONGC.NS",
    "Oil India":         "OIL.NS",
    "Vedanta":           "VEDL.NS",

    # Defence
    "HAL":               "HAL.NS",
    "BDL":               "BDL.NS",
    "BEL":               "BEL.NS",
    "BEML":              "BEML.NS",
    "Mazagon Dock":      "MAZDOCK.NS",
    "GRSE":              "GRSE.NS",
    "Cochin Shipyard":   "COCHINSHIP.NS",
    "Data Patterns":     "DATAPATTNS.NS",

    # Aviation
    "IndiGo":            "INDIGO.NS",
    "SpiceJet":          "SPICEJET.NS",
    "Air India":         "AIRINDIA.NS",

    # Paints
    "Asian Paints":      "ASIANPAINT.NS",
    "Berger Paints":     "BERGEPAINT.NS",
    "Kansai Nerolac":    "KANSAINER.NS",
    "Indigo Paints":     "INDIGOPNTS.NS",

    # Tyres
    "MRF":               "MRF.NS",
    "Apollo Tyres":      "APOLLOTYRE.NS",
    "CEAT":              "CEATLTD.NS",
    "Balkrishna Ind":    "BALKRISIND.NS",

    # Pharma
    "Sun Pharma":        "SUNPHARMA.NS",
    "Dr Reddys":         "DRREDDY.NS",
    "Divis Labs":        "DIVISLAB.NS",
    "Cipla":             "CIPLA.NS",

    # Shipping & Ports
    "Adani Ports":       "ADANIPORTS.NS",
    "Shipping Corp":     "SCI.NS",
    "Great Eastern":     "GESHIP.NS",
    "Essar Shipping":    "ESSARSHPNG.NS",

    # Capital Goods
    "L&T":               "LT.NS",
    "Thermax":           "THERMAX.NS",
    "KEC International": "KEC.NS",

    # Gold-Linked
    "Titan":             "TITAN.NS",
    "Kalyan Jewellers":  "KALYANKJIL.NS",
    "Muthoot Finance":   "MUTHOOTFIN.NS",
    "Manappuram":        "MANAPPURAM.NS",

    # Chemicals & Fertilisers
    "Coromandel":        "COROMANDEL.NS",
    "Deepak Nitrite":    "DEEPAKNTR.NS",
    "GNFC":              "GNFC.NS",
    "Tata Chemicals":    "TATACHEM.NS",

    # IT & Tech
    "Infosys":           "INFY.NS",
    "Wipro":             "WIPRO.NS",
    "HCL Tech":          "HCLTECH.NS",
}

START = "2026-02-28"
END   = "2026-04-16"

print(f"Downloading {len(tickers)} stocks...")

raw = yf.download(
    list(tickers.values()),
    start=START,
    end=END,
    interval="1d",
    auto_adjust=True,
    progress=True
)

ticker_to_name = {v: k for k, v in tickers.items()}

frames = []
for ticker, name in ticker_to_name.items():
    if ticker not in raw["Close"].columns:
        print(f"  NOT FOUND: {name} ({ticker})")
        continue
    df = pd.DataFrame({
        "date":    raw.index,
        "company": name,
        "ticker":  ticker,
        "open":    raw["Open"][ticker].values,
        "high":    raw["High"][ticker].values,
        "low":     raw["Low"][ticker].values,
        "close":   raw["Close"][ticker].values,
        "volume":  raw["Volume"][ticker].values,
    })
    frames.append(df)

long_df = pd.concat(frames, ignore_index=True)
long_df["date"] = pd.to_datetime(long_df["date"]).dt.date

print(f"\nRows: {len(long_df)}")
print(f"Companies: {long_df['company'].nunique()}")
print(f"Missing values:\n{long_df.isnull().sum()}")

long_df.to_csv("ohlc_raw.csv", index=False)
print("\nSaved: ohlc_raw.csv")

```
